##### **2a) Distribution Analysis**

````python
# Create a list of numerical features to analyse against HeartDisease.

numerical_features = [
    'Age',
    'MaxHR',
    'Oldpeak',
    'RestingBP',
    'Cholesterol'
]


# Loop through each numerical feature.

for col in numerical_features:

    # Create a boxplot showing the distribution of the numerical feature
    # for patients with and without heart disease.

    ax = sns.boxplot(
          data=clean,
          x='HeartDisease',
          y=col,
          hue='HeartDisease',
          palette='viridis',
          legend=False
        )

    # Add a title to the chart.
    plt.title(f'{col} Distribution by Heart Disease')

    # Add a label to the x-axis.
    plt.xlabel('Heart Disease')

    # Add a label to the y-axis.
    plt.ylabel(col)

    # Change the x-axis labels to more meaningful names.
    plt.xticks(
         [0, 1],
         ['No Heart Disease', 'Heart Disease']
    )

    # Remove the top and right borders for a cleaner look.
    sns.despine(top=True, right=True)

    # Display the plot.
    plt.show()
````

##### **Interpretation**

These boxplots visualize the distribution of key numerical features across patients with and without heart disease, offering insights into how these variables differentiate the two groups.

*   **Age Distribution by Heart Disease:**
    *   Patients with heart disease (`HeartDisease = 1`) tend to be older, with a higher median age and a box shifted towards higher values compared to those without heart disease (`HeartDisease = 0`).
    *   The interquartile range (IQR) for patients with heart disease also appears to be slightly wider, indicating more variability in age within this group.

*   **MaxHR Distribution by Heart Disease:**
    *   Patients with heart disease generally exhibit a lower maximum heart rate (MaxHR). The median MaxHR for the heart disease group is noticeably lower than for the group without heart disease.
    *   The spread of MaxHR values is similar between the two groups, but the entire distribution for heart disease patients is shifted downwards.

*   **Oldpeak Distribution by Heart Disease:**
    *   Patients with heart disease show significantly higher Oldpeak values, with a median that is much greater than that of the group without heart disease.
    *   The box for the heart disease group is also much wider and higher, indicating greater variability and a tendency for more pronounced ST depression in these patients. There are also more apparent outliers with very high Oldpeak values in the heart disease group.

*   **RestingBP Distribution by Heart Disease:**
    *   The distributions of RestingBP for both groups are quite similar, with overlapping medians and interquartile ranges. There isn't a strong visual distinction based on RestingBP alone.
    *   Both groups show some outliers with higher resting blood pressure, but there's no clear pattern that strongly separates heart disease patients from healthy ones based on this feature's distribution.

*   **Cholesterol Distribution by Heart Disease:**
    *   Similar to RestingBP, the distributions of Cholesterol levels for both groups largely overlap. The medians are very close, and the boxes show similar spreads.

##### **b) Reconstruct ChestPainType**

````python
# Reconstruct ChestPainType

def get_ChestPainType(row):
    if row["ChestPainType_ATA"] == 1:
        return "Atypical Angina"
    elif row["ChestPainType_NAP"] == 1:
        return "Non-Anginal Pain"
    elif row["ChestPainType_TA"] == 1:
        return "Typical Angina"
    else:
        return "Asymptomatic"


week3["ChestPainType"] = week3.apply(
    get_ChestPainType,
    axis=1
)


# Check the reconstructed categories
print(week3["ChestPainType"].value_counts())
````

##### **ChestPainType by HeartDisease**

````python
# Count of patients with and without ChestPainType,
# grouped by heart disease status

ax = sns.countplot(
    data=week3,
    x="ChestPainType",
    hue="HeartDisease",
    palette="viridis",
    order = (week3["ChestPainType"].value_counts(ascending=False).index
))

# Add title
plt.title("Chest Pain Type by Heart Disease")

# Add axis labels
plt.xlabel("Chest Pain Type")
plt.ylabel("Count")

# Add count values to the bars
for container in ax.containers:
    ax.bar_label(container)

# Improve legend labels
ax.legend(
    title="Heart Disease",
    labels=["Normal", "Heart Disease"]
)

# Remove top and right borders
sns.despine(top=True, right=True)

# Display plot
plt.show()
````

##### **Percentage Distribution of Heart Disease by Chest_Pain_Type**

````python
chest_pain_pct = pd.crosstab(
    week3["ChestPainType"],
    week3["HeartDisease"],
    normalize="index"
) * 100

chest_pain_pct.columns = [
    "Normal Disease",
    "Heart Disease"
]

chest_pain_pct
````

##### **1) Business question**
*   Is Age significantly different between patients with and without heart disease?

**Objective of the Test**
*   To determine whether the observed age difference is statistically significant.

**Hypothesis Testing**

**Null hypothesis (H₀)**
*   There is no statistically significant difference in Age between the two heart-disease groups.

**Alternate hypothesis (H₁)**
*   There is a statistically significant difference in Age between the two groups.

**Justification**
*   The Mann–Whitney U test was selected because Age did not meet the **assumption of normality (p < 0.05)**, and the analysis involves comparing Age between two independent groups—patients with and without heart disease.
*   As a non-parametric test, the Mann–Whitney U test is appropriate for determining whether the Age distributions differ significantly between the two groups without assuming normality.

##### **Statistical Test: Age vs HeartDisease(Mann–Whitney U)**

````python
from scipy.stats import mannwhitneyu

# Select the Age values for patients without heart disease (0).
group0 = week3.loc[
    week3["HeartDisease"] == 0,
    "Age"
]

# Select the Age values for patients with heart disease (1).
group1 = week3.loc[
    week3["HeartDisease"] == 1,
    "Age"
]

# Perform a two-sided Mann-Whitney U test to determine whether
# Age differs significantly between the two HeartDisease groups.
stat, p = mannwhitneyu(
    group0,
    group1,
    alternative="two-sided"
)

# Display the Mann-Whitney U test statistic.
print("Statistic:", stat)

# Display the p-value used to determine statistical significance.
print("p-value:", p)
````

##### **Decision**

````python
# Set the significance level at 5%.
alpha = 0.05

# Compare the p-value with the significance level to determine
# whether the null hypothesis should be rejected.
if p < alpha:

    # A p-value below 0.05 indicates a statistically significant result.
    print("Reject H0")

else:
    # A p-value of 0.05 or greater indicates insufficient evidence
    # to reject the null hypothesis.

    print("Fail to reject H0")
````

**Conclusion**
*   The Mann–Whitney U test identified a statistically significant difference in Age between patients with and without heart disease **(U = 69,137.5, p < 0.001)**.
*   Therefore, we reject the null hypothesis, and conclude that there statistically significant difference in Age between the two groups.
*   Thus,Older age is significantly associated with heart-disease status.


**Business implication**
*   The significant difference in age between patients with and without heart disease suggests that age is an important factor when assessing heart disease risk.
*   Healthcare providers could use age as part of an initial risk-stratification process to identify individuals who may benefit from closer monitoring or further cardiovascular assessment.
*   However, age should be considered alongside other clinical factors rather than used as a standalone predictor.

**ST_Slope vs HeartDisease**

**4) Business question**
*   Is ST Slope statistically associated with heart disease?

**Objective of the test**
*   To determine whether there is a statistically significant association between ST_Slope and HeartDisease, and whether the distribution of ST slope categories differs between individuals with and without heart disease.

**Hypothesis Testing**

**Null hypothesis (H₀)**
*   ST Slope and HeartDisease are independent.

**Alternate hypothesis (H₁)**

*   ST Slope and HeartDisease are not independent.

**Chi-Square Test of Independence**

**Justification**
*    The Chi-square test of independence was used to examine the relationship between ST_Slope and HeartDisease because both variables are categorical.

*    The test determines whether the distribution of ST_Slope categories differs significantly according to heart disease status.

````python
from scipy.stats import chi2_contingency

# Create a contingency table showing the frequency of HeartDisease
# for each ST_Slope category.
table = pd.crosstab(
    week3["ST_Slope"],
    week3["HeartDisease"]
)

# Perform a Chi-square test of independence to determine whether
# ST_Slope and HeartDisease are significantly associated.
chi2, p, dof, expected = chi2_contingency(table)

# Display the Chi-square test statistic.
print("Chi-square:", chi2)

# Display the p-value used to determine statistical significance.
print("p-value:", p)

# Display the degrees of freedom for the test.
print("Degrees of freedom:", dof)
````

**Conclusion**
*    The results showed a statistically significant association between the two variables,**(χ2(2)=355.918, p<0.001).**
*    Therefore, we reject the null hypothesis and conclude that ST_Slope is significantly associated with heart disease status.

**Business implication**
*    The strong association suggests that ST-slope characteristics could be useful when assessing a patient's likelihood of heart disease.

*    Healthcare providers could consider ST_Slope alongside other clinical indicators when developing risk-assessment or screening tools.

````python
from sklearn.feature_selection import mutual_info_classif

# Recreate X using the 'clean' DataFrame (which has original numerical/boolean columns)
# and add the newly engineered numerical feature 'Heart_Rate_Age_Ratio' from week3.
X_mi = clean.drop(columns=["HeartDisease"]).copy()
X_mi["Heart_Rate_Age_Ratio"] = week3["Heart_Rate_Age_Ratio"]

y = week3["HeartDisease"]

# Identify discrete features for mutual_info_classif.
# Boolean columns and FastingBS (originally int64 0/1) are discrete.
# Other numerical columns are continuous.
feature_names = X_mi.columns
discrete_features_mask = [
    (X_mi[col].dtype == 'bool') or (col == 'FastingBS')
    for col in feature_names
]

mi = mutual_info_classif(X_mi, y, discrete_features=discrete_features_mask, random_state=42)
````

````python
mi_series = pd.Series(mi, index=X_mi.columns).sort_values(ascending=False)
display(mi_series)
````

##### **Interpretation**

Mutual Information (MI) measures the dependency between variables, indicating how much knowing one variable reduces uncertainty about the other. Higher MI scores suggest a stronger relationship.

*   **Highest MI Scores**: Features like `ST_Slope_Flat`, `ST_Slope_Up`, `ExerciseAngina_Y`, `Oldpeak`, and `MaxHR` show the highest mutual information scores with `HeartDisease`. This indicates these features are individually strong predictors and provide the most information about the target variable.
*   **Moderate MI Scores**: `Age`, `Cholesterol`, and `RestingBP` show moderate MI scores, suggesting they still hold valuable information regarding heart disease, although perhaps less directly or in combination with other features.
*   **Lower MI Scores**: Features derived from one-hot encoding with fewer occurrences (e.g., `ChestPainType_TA`, `RestingECG_ST`, `RestingECG_Normal`) tend to have lower MI scores. This might be due to their binary nature or the distribution of their categories.
*   **Overall**: The MI scores generally align with the insights from statistical tests and correlation analysis, confirming the predictive power of key features, including the engineered `Heart_Rate_Age_Ratio`.

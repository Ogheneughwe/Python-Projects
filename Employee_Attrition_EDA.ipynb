### **Univariate Analysis**

### **Understand the Target Variable**

````python
# Attrition
df['Attrition'].value_counts()
````

````python
plt.figure(figsize=(8, 8))
colors = [sns.color_palette('viridis')[1], sns.color_palette('viridis')[3]] # Using a darker blue from viridis for 'No' and a viridis color for 'Yes'

# Get the value counts for Attrition
attrition_counts = df['Attrition'].value_counts()
labels_data = attrition_counts.index.tolist() # e.g., ['No', 'Yes']
sizes = attrition_counts.values.tolist() # e.g., [1233, 237]
total = sum(sizes)

# Create custom labels with value and percentages in brackets
custom_labels = []
for i, label_text in enumerate(labels_data):
    percentage = (sizes[i] / total) * 100
    custom_labels.append(f"{label_text} ({sizes[i]}, {percentage:.1f}%)")

# Plot the pie chart using matplotlib.pyplot.pie for more control over labels
wedges, texts = plt.pie(
    sizes,
    labels=custom_labels,  # Use our custom labels
    colors=colors,
    startangle=90,
    wedgeprops={'edgecolor': 'white'},
    labeldistance=1.1 # Position labels slightly outside the wedges
)

plt.title('Attrition Distribution')
plt.ylabel('') # Hide the y-label for a cleaner pie chart
plt.show()
````

##### **Interpretation**

We can observe that:

*    `1233` employees who did not leave the company.
*    `237` employees left the company.



##### **Conclusion**

*   A significant larger number of employees stayed with the company compared to those who left.

*   The dataset is **imbalanced** with
respect to the target variable 'Attrition'.

### **6. Which job roles experience the highest turnover?**

````python
# Calculate attrition counts per JobRole
attrition_by_jobrole = df.groupby(['JobRole', 'Attrition']).size().unstack(fill_value=0)

# Calculate total employees per JobRole
attrition_by_jobrole['Total'] = attrition_by_jobrole['No'] + attrition_by_jobrole['Yes']

# Calculate attrition rate
attrition_by_jobrole['Attrition_Rate'] = (attrition_by_jobrole['Yes'] / attrition_by_jobrole['Total']) * 100

# Sort by Attrition Rate in descending order
attrition_by_jobrole_sorted = attrition_by_jobrole.sort_values(by='Attrition_Rate', ascending=False)

display(attrition_by_jobrole_sorted)
````

### **7. Which variables appear important for future predictive modelling?**

````python
# Create a copy of the DataFrame to avoid modifying the original 'df' for this specific task.
df_corr = df.copy()

# Convert 'Attrition' to numerical (Yes=1, No=0)
df_corr['Attrition'] = df_corr['Attrition'].map({'Yes': 1, 'No': 0})

# Drop columns that have only one unique value as they will not contribute to correlation.
df_corr = df_corr.drop(columns=['EmployeeCount', 'StandardHours', 'Over18'], errors='ignore')

# Drop other categorical columns that are not yet numerically encoded and would cause errors
# or are not directly relevant for a numerical correlation matrix.
df_corr = df_corr.drop(columns=[
    'BusinessTravel', 'Department', 'EducationField', 'Gender', 'JobRole',
    'MaritalStatus', 'OverTime', 'Age_Group', 'Education_Group',
    'EnvironmentSatisfaction_Group', 'JobSatisfaction_Group', 'JobInvolvement_Group',
    'PerformanceRating_Group', 'RelationshipSatisfaction_Group', 'WorkLifeBalance_Group'
], errors='ignore')

# Calculate the correlation matrix for the cleaned numerical data
correlation_matrix = df_corr.corr()

# Display the correlation matrix for 'Attrition' with other features
# Sort by absolute correlation value to easily see the strongest relationships
display(correlation_matrix['Attrition'].sort_values(ascending=False).to_frame())
````

Based on the comprehensive analysis of the IBM HR Analytics dataset, the following recommendations are proposed to address employee attrition at ABC Manufacturing Ltd:

##### 1. **Target Retention Programs for Younger Employees**
*   **Insight**: Younger employees, particularly those in the 18-25 Young Adult and 26-35 Early Career age groups, exhibit significantly higher attrition rates.
*   **Recommendation**: Develop tailored mentorship programs, career development opportunities, and engagement initiatives specifically designed for early-career professionals. Ensure clear growth paths and regular feedback to address their career aspirations and reduce turnover.

##### 2. **Review Compensation and Benefits for Lower-Income Roles**
*   **Insight**: Employees with lower monthly incomes are significantly more prone to attrition.
*   **Recommendation**: Conduct a review of salary structures for positions with lower monthly incomes. Consider increasing base salaries, offering performance-based bonuses, or enhancing benefit packages (e.g., healthcare, retirement plans) to improve financial satisfaction and reduce the incentive to seek higher-paying opportunities elsewhere.

##### 3. **Address Overtime Culture and Work-Life Balance**
*   **Insight**: Working overtime is strongly associated with a higher likelihood of attrition.
*   **Recommendation**: Implement policies to monitor and reduce excessive overtime. Promote a healthy work-life balance through flexible work arrangements, sufficient staffing to alleviate workload, and initiatives that encourage employees to disconnect outside of working hours. Burnout prevention programs could also be beneficial.

##### 4. **Investigate and Support High-Turnover Job Roles**
*   **Insight**: Specific job roles, such as Sales Representatives (39.76% attrition), Laboratory Technicians (23.94%), and Human Resources (23.08%), experience notably higher turnover rates.
*   **Recommendation**: Conduct in-depth departmental-level analyses for these high-attrition roles to understand specific pain points. This could involve exit interviews, employee satisfaction surveys, and focus groups. Based on findings, implement targeted improvements in working conditions, management support, training, and career progression for these roles.

##### 5. **Enhance Employee Engagement and Satisfaction Metrics**
*   **Insight**: Lower `TotalWorkingYears`, `JobLevel`, `YearsInCurrentRole`, `JobSatisfaction`, and `EnvironmentSatisfaction` are negatively correlated with attrition, indicating that employees with less experience/tenure and lower satisfaction are more likely to leave.
*   **Recommendation**: Focus on initiatives that improve overall job satisfaction, involvement, and environmental satisfaction. This includes fostering a positive work environment, promoting inclusive workplace culture, providing opportunities for skill development and advancement, and ensuring fair and transparent promotion processes. Regularly measure these satisfaction metrics and act on the feedback.

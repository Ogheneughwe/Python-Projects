### **Description of Variables**

The dataset contains the following key variables related to employee information and attrition:

*   **Age**: Age of the employee.
*   **Attrition**: Whether the employee left the company (Yes) or not (No). This is the target variable.
*   **BusinessTravel**: Frequency of business travel for the employee (e.g., Travel_Rarely, Travel_Frequently, Non-Travel).
*   **DailyRate**: Daily rate of pay for the employee.
*   **Department**: Department the employee works in (e.g., Sales, Research & Development, Human Resources).
*   **DistanceFromHome**: Distance from home to workplace in miles.
*   **Education**: Level of education (1 'Below College' to 5 'Doctor').
*   **EducationField**: Field of education (e.g., Life Sciences, Medical, Marketing).
*   **EmployeeCount**: Number of employees (always 1 for each row, can be dropped).
*   **EmployeeNumber**: Unique employee identification number.
*   **EnvironmentSatisfaction**: Satisfaction with the work environment (1 'Low' to 4 'Very High').
*   **Gender**: Gender of the employee (Male, Female).
*   **HourlyRate**: Hourly rate of pay.
*   **JobInvolvement**: Level of job involvement (1 'Low' to 4 'Very High').
*   **JobLevel**: Job level of the employee (1 to 5).
*   **JobRole**: Role of the employee (e.g., Sales Executive, Research Scientist).
*   **JobSatisfaction**: Satisfaction with the job (1 'Low' to 4 'Very High').
*   **MaritalStatus**: Marital status (e.g., Single, Married, Divorced).
*   **MonthlyIncome**: Monthly income of the employee.
*   **MonthlyRate**: Monthly rate of pay.
*   **NumCompaniesWorked**: Number of companies worked for prior to IBM.
*   **Over18**: Whether the employee is over 18 (all 'Y', can be dropped).
*   **OverTime**: Whether the employee works overtime (Yes, No).
*   **PercentSalaryHike**: Percentage increase in salary.
*   **PerformanceRating**: Performance rating (1 'Low' to 4 'Outstanding').
*   **RelationshipSatisfaction**: Satisfaction with relationships at work (1 'Low' to 4 'Very High').
*   **StandardHours**: Standard hours of work (always 80, can be dropped).
*   **StockOptionLevel**: Level of stock option (0 to 3).
*   **TotalWorkingYears**: Total number of years worked.
*   **TrainingTimesLastYear**: Number of training times last year.
*   **WorkLifeBalance**: Work-life balance (1 'Bad' to 4 'Best').
*   **YearsAtCompany**: Total number of years at the company.
*   **YearsInCurrentRole**: Number of years in the current role.
*   **YearsSinceLastPromotion**: Number of years since the last promotion.
*   **YearsWithCurrManager**: Number of years with current manager.
